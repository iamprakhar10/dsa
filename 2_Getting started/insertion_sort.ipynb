```python
def insertion_sort(arr):
    l = len(arr)
    i=1
    while i<l:
        current = i
        while current-1>=0 and arr[current]<arr[current-1]:
            arr[current-1], arr[current] = arr[current], arr[current-1]
            current -= 1
        i += 1
    print(arr)
```

```python
insertion_sort([5,2,4,6,1,3])
```

```python
insertion_sort([2,2])
```

```python
insertion_sort([2])
```

```python
def insertion_sort_2(arr):
    for j in range(1, len(arr)):
        key = arr[j]
        while j-1>=0 and arr[j-1]>key:
            arr[j] = arr[j-1]
            j-=1
        arr[j] = key
    print(arr)
```

```python
insertion_sort_2([5,2,4,6,1,3])
```

```python
insertion_sort_2([1,2,1])
```

We can express insertion sort as a recursive procedure as follows. In order to sort
A[1 ...n], we recursively sort A[1 ...n-1] and then insert A[n] into the sorted array
A[1 ...n-1]. Write a recurrence for the running time of this recursive version of
insertion sort.

```python
def merge_insertion(arr, p, q):
    key = arr[q]
    q -= 1
    while q>=p:
        if arr[q]>key:
            arr[q+1] = arr[q]
            q -= 1
        else:
            break
    arr[q+1] = key
    print(arr)

def insertion_sort_recursive(arr, p, q):
    if q>p:
        insertion_sort_recursive(arr, p, q-1)
        merge_insertion(arr, p, q)
```

 if the
sequence A is sorted, we can check the midpoint of the sequence against and
eliminate half of the sequence from further consideration. The binary search algorithm repeats this procedure, halving the size of the remaining portion of the
sequence each time. Write pseudocode, either iterative or recursive, for binary
search. Argue that the worst-case running time of binary search is ‚O(lg n)/.

```python
def binary_search(arr, p, r, target):
    if p<=r:
        q = (p+r)//2
        if arr[q]==target:
            return q
        elif arr[q]<target:
            return binary_search(arr, q+1, r, target)
        else:
            return binary_search(arr, p, q-1, target)
```

```python
arr = list(range(13))
for i in range(13):
    print(binary_search(arr, 0, len(arr)-1, i))
    print(binary_search(arr, 0, len(arr)-1, i+.8))
```
