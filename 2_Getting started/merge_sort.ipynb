```python
def merge(arr, p, q, r):
    """
    a -> p
    p+1 -> r
    """
    L = arr[p:q+1]
    R = arr[q+1:r+1]
    L.append(float("inf"))
    R.append(float("inf"))

    i,j = 0,0
    for k in range(p, r+1):
        if L[i]<R[j]:
            arr[k] = L[i]
            i+=1
        else:
            arr[k] = R[j]
            j+=1



def merge_sort(arr, p, r):
    if p < r:
        q = (p+r)//2
        merge_sort(arr, p, q)
        merge_sort(arr, q+1, r)
        merge(arr, p , q, r)
```

```python
arr = [5,2,4,6,1,3]
merge_sort(arr, 0, 5)
arr
```
