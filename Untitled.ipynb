```python
def threeSum(nums: list[int]) -> list[list[int]]:
        d = {}
        s=0
        res = []
        nums = sorted(nums)
    
        arr = [nums[0], nums[1], nums[2]]
        s1 = 3
        t = []
        while s1 < len(nums) and nums[s1] != nums[s1-3]:
            arr.append(nums[s1])
            s1 += 1
        nums=arr
        for i in nums:
            if i not in d:
                d[i] = [s]
            else:
                d[i].append(s)
            s += 1
        for i in range(s-2):
            for j in range(i+1, s-1):
                a = nums[i] + nums[j]
                if -a in d:
                    for k in d[-a]:
                        if k not in (i,j) and k>j:
                            res.append([nums[i], nums[j],nums[k]])
        a = [sorted(r) for r in res]
        ans = []
        for i in a:
            if i not in ans:
                ans.append(i)
        return ans
```
