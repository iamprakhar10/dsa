```python
def maximum_between_interval(arr, p, r, m = -float("inf")):

    print(p,r,m)
    if r>=len(arr):
        return -float('inf')
    q = (p+r)//2
    if p==r-1:
        m = max(m, arr[r]-arr[p])
        return m
    if p == r:
        return -float('inf')
    m =  max(maximum_between_interval(arr,p,q,m), 
            maximum_between_interval(arr, q+1,r, m),
            max(arr[q+1:r+1])-min(arr[p:q+1]))
    return m
```

```python
arr = [100,113,110,85,105,102,86,63,81,101,94,106,101,79,94,90,97]
maximum_between_interval(arr, 0, len(arr)-1)
```
